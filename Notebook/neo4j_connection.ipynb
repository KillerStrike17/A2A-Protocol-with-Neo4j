```python
# Neo4j public read-only demo database (Recommendations dataset), see https://demo.neo4jlabs.com
# Replace with your own Neo4j Sandbox / AuraDB / local credentials if you prefer.
NEO4J_URI="neo4j+s://demo.neo4jlabs.com"
NEO4J_USERNAME="recommendations"
NEO4J_PASSWORD="recommendations"
NEO4J_DATABASE="recommendations"
```

```python
from neo4j import GraphDatabase

# URI examples: "neo4j://localhost", "neo4j+s://xxx.databases.neo4j.io"
URI = NEO4J_URI
AUTH = (NEO4J_USERNAME, NEO4J_PASSWORD)

# Keep the driver open for the cells below; it is closed in the last cell.
driver = GraphDatabase.driver(URI, auth=AUTH)
driver.verify_connectivity()
print("Connected to Neo4j")
```

```python
records, summary, keys = driver.execute_query("""
    MATCH (n) RETURN n LIMIT 25;
    """,
    database_=NEO4J_DATABASE,
)
```

```python
# Loop through results and do something with them
for record in records:
    print(record.data())  # obtain record as dict

# Summary information
print("The query `{query}` returned {records_count} records in {time} ms.".format(
    query=summary.query, records_count=len(records),
    time=summary.result_available_after
))
```

```python
records, summary, keys = driver.execute_query("""
    MATCH (m:Movie)-[:IN_GENRE]->(g:Genre {name: $genre})
    WHERE m.imdbRating IS NOT NULL
    RETURN m.title AS title, m.year AS year, m.imdbRating AS rating
    ORDER BY m.imdbRating DESC
    LIMIT 25
    """,
    genre="Horror",
    database_=NEO4J_DATABASE,
)
```

```python
# Loop through results and do something with them
for record in records:
    print(record.data())  # obtain record as dict

# Summary information
print("The query `{query}` returned {records_count} records in {time} ms.".format(
    query=summary.query, records_count=len(records),
    time=summary.result_available_after
))
```

```python
# Close the driver when you're done
driver.close()
```
