```python
# Top Movies

MATCH (m:Movie)
WHERE m.imdbRating IS NOT NULL
RETURN m.title, m.imdbRating
ORDER BY m.imdbRating DESC
LIMIT 10;
```

```python
# Movie Genre

MATCH (m:Movie)-[:IN_GENRE]->(g:Genre)
RETURN g.name AS Genre, count(m) AS Movies
ORDER BY Movies DESC;
```

```python
# MOvies released in a Year

MATCH (m:Movie)
RETURN m.year AS Year, count(*) AS Movies
ORDER BY Year ASC;
```

```python
# Most active actor

MATCH (a:Actor)-[:ACTED_IN]->(m:Movie)
RETURN a.name AS Actor, count(m) AS Movies
ORDER BY Movies DESC
LIMIT 10;
```

```python
# MOst active director

MATCH (d:Director)-[:DIRECTED]->(m:Movie)
RETURN d.name AS Director, count(m) AS Movies
ORDER BY Movies DESC
LIMIT 10;
```

```python
# Actors and Directors

MATCH (a:Actor)-[:ACTED_IN]->(m:Movie)<-[:DIRECTED]-(d:Director)
RETURN m.title, collect(a.name) AS Actors, d.name AS Director
LIMIT 10;
```

```python
# Users who rated most movies

MATCH (u:User)-[r:RATED]->(m:Movie)
RETURN u.userId AS user, count(r) AS Ratings
ORDER BY Ratings DESC
LIMIT 10;
```

```python
# Rating Distribution

MATCH (:User)-[r:RATED]->(m:Movie)
RETURN r.rating AS Rating, count(*) AS Count
ORDER BY Rating;
```

```python
# Duplicate Movie check

MATCH (m:Movie)
WITH m.title AS title, m.year AS year, collect(m) AS movies
WHERE size(movies) > 1
RETURN title, year, size(movies) AS duplicates;
```
